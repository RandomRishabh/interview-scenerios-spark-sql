```python
from pyspark import *
from pyspark.sql import *
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import *
```

```python
data = [(1, 101, '2024-01-01', 500),(2, 102, '2024-01-01', 300),(3, 101, '2024-01-05', 700),(4, 103, '2024-01-07', 200),(5, 102, '2024-01-08', 100),(6, 101, '2024-01-09', 900)]
```

```python
df = spark.createDataFrame(data,["order_id","customer_id","order_date","amount"])
df.show()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window 
result_df= (df
            .withColumn(
                "rank", 
                F.row_number().over(
                    Window.partitionBy("customer_id")
                    .orderBy(F.col("order_date").desc())
                    )
                )
            .filter(F.col("rank")==1)
            .drop("rank")
            .select("customer_id","order_id","order_date","amount")
            )
result_df.show()
```
