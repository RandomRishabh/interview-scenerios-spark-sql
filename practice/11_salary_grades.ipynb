# 11. Salary grades

**Easy**

Add grade C below 5000, B from 5000 through 10000 inclusive, and A above 10000. Keep all input columns.


```python
%run ./_setup $exercise_id="11"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .withColumn(
            "grade",
            F.when(F.col("salary") < 5000 ,"C")
            .when((F.col("salary")<=10000) & (F.col("salary")>=5000) ,"B")
            .otherwise("A")
            )
    )

solve(input_df).show()
```
