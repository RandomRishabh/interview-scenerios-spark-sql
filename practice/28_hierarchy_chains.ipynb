# 28. Child parent grandparent chains

**Medium**

Return child, parent, grandparent for every two-edge chain in input_df(child,parent). Omit children whose parent has no parent row. Input edges are unique.


```python
%run ./_setup $exercise_id="28"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.
    return (
        input_df.alias("a")
        .join(input_df.alias("b"),F.col("a.parent")==F.col("b.child"))
        # .withColumn("grandparent",F.col("b.parent"))
        .filter(F.col("b.parent").isNotNull())
        .select(
            F.col("a.child"),
            F.col("a.parent"),
            F.col("b.parent").alias("grandparent")
        )
    )

solve(input_df).display()
```
