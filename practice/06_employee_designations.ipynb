```python
from pyspark.sql import functions as F

def solve(input_df):
    
    return (
        input_df
        .withColumn(
            "Designation", 
            F.when(F.col("salary")>10000, "Manager")
            .otherwise("Employee")
        )
    )
solve(input_df).show()
```
