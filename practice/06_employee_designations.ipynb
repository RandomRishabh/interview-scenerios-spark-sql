# 06. Assign employee designations

**Conditional transformations · Easy**

Keep all input columns and add Designation: Manager for a numeric salary greater than 10000, otherwise Employee. salary is supplied as a string; preserve its output type. Use the correctly spelled output name Designation.


```python
%run ./_setup $exercise_id="6"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    
    return (
        input_df
        .withColumn(
            "Designation", 
            F.when(F.col("salary")>10000, "Manager")
            .otherwise("Employee")
        )
    )
solve(input_df).show()
```
