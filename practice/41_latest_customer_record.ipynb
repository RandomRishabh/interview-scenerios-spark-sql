# 41. Latest customer record after CDC

**Medium**

Keep one latest row per customer_id by updated_at, supplied as sortable ISO timestamp strings. For an equal timestamp choose city ascending, then customer_name ascending; these fields are non-null. Exact duplicate input rows still produce one output row. Keep the four input columns. Input fixtures use customer records, correcting the upstream README copied login table.

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .withColumn("rank", F.row_number().over(Window.partitionBy("customer_id").orderBy(F.col("updated_at").desc(),F.col("city"),F.col("customer_name"))))
        .filter(F.col("rank")==1)
        .drop("rank")
    )

solve(input_df).display()
```
