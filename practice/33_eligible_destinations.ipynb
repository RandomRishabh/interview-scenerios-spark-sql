# 33. Families with most eligible destinations

**Hard**

A family qualifies for a country when family_size is between min_size and max_size inclusive. Return name and number_of_countries for all families tied for the largest positive number of eligible countries. Family names are unique. If no family qualifies anywhere, return empty.


```python
%run ./_setup $exercise_id="33"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(families, countries):
    # Write your solution here.
    return (
        families.alias("a")
        .crossJoin(countries.alias("b"))
        .filter((F.col("family_size") <= F.col("max_size")) & (F.col("family_size") >= F.col("min_size")))
        .groupBy("a.name")
        .agg(F.count("a.id").alias("number_of_countries"))
        .orderBy(F.col("number_of_countries").desc())
        .withColumn("max", F.max("number_of_countries").over(Window.partitionBy()))
        .filter(F.col("number_of_countries")>1)
        .filter(F.col("number_of_countries")== F.col("max"))
        .drop("max")
    )

solve(families,countries).display()
```
