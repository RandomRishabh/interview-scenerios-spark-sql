# 19. Flatten nested feed events

**Hard**

Return one output row for each element of multiMedia; records with an empty/null media array produce no rows. Flatten likeDislike into dislikes, likes, userAction and media fields into top-level columns. On collisions, media description/id/mediatype/name replace the parent fields. Keep all other parent columns and drop likeDislike/multiMedia. Inspect the expected schema.


```python
%run ./_setup $exercise_id="19"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.
    result_df=(
        input_df
        .withColumn("media",F.explode("multiMedia"))
        .withColumn("dislikes",F.col("likeDislike.dislikes"))
        .withColumn("likes",F.col("likeDislike.likes"))
        .withColumn("userAction",F.col("likeDislike.userAction"))
        .withColumn("createAt",F.col("media.createAt"))
        .withColumn("likeCount",F.col("media.likeCount"))
        .withColumn("place",F.col("media.place"))
        .withColumn("url",F.col("media.url"))
        .withColumn("description", F.col("media.description"))
        .withColumn("id", F.col("media.id"))
        .withColumn("mediatype", F.col("media.mediatype"))
        .withColumn("name", F.col("media.name"))
        .drop("likedislike","media","multimedia")
    )
    return result_df

solve(input_df).display()
```
