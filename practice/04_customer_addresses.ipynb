# 04. Distinct addresses per customer

**Medium**

Return custid, custname, and address as an array of distinct addresses per customer. Ignore null addresses. Address array order does not matter. Despite the original prose mentioning counts, this contract matches its array output.


```python
%run ./_setup $exercise_id="4"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    result_df= (
        input_df
        .groupBy("custid","custname")
        .agg(F.collect_set("address").alias("address"))
    )
    # Write your solution here.
    return result_df
solve(input_df).show()
```
