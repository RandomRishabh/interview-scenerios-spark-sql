```python
from pyspark.sql import functions as F

def solve(input_df):
    result_df= (
        input_df
        .groupBy("custid","custname")
        .agg(F.collect_set("address").alias("address"))
    )
    # Write your solution here.
    return result_df
solve(input_df).show()
```
