# 09. Participants with the most first places

**Arrays and aggregation · Medium**

Count rank=1 occurrences for each participant and return the name column for every participant tied for the largest positive count. If nobody has a first place, return an empty DataFrame. Return a DataFrame even though the reference prints a scalar.


```python
%run ./_setup $exercise_id="9"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.
    result=(
        input_df
        .withColumn("item",F.explode("rank"))
        .filter(F.col("item")==1)
        .groupBy("name")
        .agg(F.count("*").alias("count"))
        .withColumn("max_count",F.max("count").over(Window.partitionBy()))
        .filter(F.col("count")==F.col("max_count"))
        .select("name")        
    )
    return result

solve(input_df).show()
```
