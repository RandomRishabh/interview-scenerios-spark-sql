```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.
    result=(
        input_df
        .withColumn("item",F.explode("rank"))
        .filter(F.col("item")==1)
        .groupBy("name")
        .agg(F.count("*").alias("count"))
        .withColumn("max_count",F.max("count").over(Window.partitionBy()))
        .filter(F.col("count")==F.col("max_count"))
        .select("name")        
    )
    return result

solve(input_df).show()
```
