# 29. Combine IDs except the first set maximum

**Medium**

Return distinct IDs from either df1.col or df2.col1, excluding the maximum ID found in df1. Inputs contain non-null integers. If df1 is empty, retain all df2 IDs. Output column: col.


```python
%run ./_setup $exercise_id="29"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(df1, df2):
    # Write your solution here.
    max_df1=df1.agg(F.max("col").alias("max"))
    result_df= (
        df1
        .union(df2)
        .distinct()
        .crossJoin(max_df1)
        .filter((F.col("col")!=F.col("max"))  | (F.col("max").isNull()))
        .select("col").orderBy("col")            
    )



    return result_df

solve(df1,df2).display()
```
