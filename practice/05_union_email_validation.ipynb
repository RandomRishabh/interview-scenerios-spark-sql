```python
from pyspark.sql import functions as F

def solve(df1, df2):
    df1= (df1
          .withColumn("salary",F.lit(1000))
          )
    result= (
          df1
          .unionByName(df2)
          .filter(F.col("email").rlike(r"@"))
            )
    return result
    # Write your solution here.
solve(df1,df2).show()
```
