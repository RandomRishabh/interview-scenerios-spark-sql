Add salary=1000 to df1, append df2, and retain rows whose non-null email contains @. Return id, name, age, email, salary. Optional ungraded Spark tasks: inspect observed partition IDs; explain partitionBy(salary) when writing. No filesystem writes are required for the checked task.

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(df1, df2):
    df1= (df1
          .withColumn("salary",F.lit(1000))
          )
    result= (
          df1
          .unionByName(df2)
          .filter(F.col("email").rlike(r"@"))
            )
    return result
    # Write your solution here.
solve(df1,df2).show()
```
