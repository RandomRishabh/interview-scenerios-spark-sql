# 01. Employees sharing a salary

**Medium**

Return each worker whose salary is shared by another worker. Return a worker once even when three or more workers share that salary. Keep all input columns.


```python
%run ./_setup $exercise_id="1"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.

    return (
        input_df
        .withColumn("count", F.count("salary").over(Window.partitionBy("salary")))
        .filter(F.col("count")>1)
        .drop("count")
    )

solve(input_df).display()
```
