# 32. Render star ratings

**Easy**

Match foods with their ratings using food_id, keeping only IDs present in both datasets, and add stats(out of 5) containing rating asterisks. Ratings are integers from 0 to 5; IDs are unique on each side. Return food_id, food_item, rating and stats(out of 5).


```python
%run ./_setup $exercise_id="32"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(foods, ratings):
    # Write your solution here.
    result_df=(
        foods.alias("a")
        .join(ratings.alias("b"),"food_id","inner")
        .withColumn("stats(out of 5)", F.repeat(F.lit("*"),F.col("rating")))
    )

    return result_df

solve(foods,ratings).display()
```
