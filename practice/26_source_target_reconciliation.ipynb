# 26. Reconcile source and target records

**Hard**

Compare source_df(id,name) and target_df(id1,name1) by unique non-null ID. Omit equal rows, treating two null names as equal. Return id and comment: new in source for a missing target ID; new in target for a missing source ID; mismatch for different names including null versus non-null. Detect missing IDs, not missing names.


```python
%run ./_setup $exercise_id="26"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(source_df, target_df):
    # Write your solution here.

    return (
        source_df.alias("a")
        .join(target_df.alias("b"),F.col("a.id")==F.col("b.id1"),"outer")
        .withColumn(
            "comment", 
            F.when(F.col("a.id").isNull(), "new in target")
            .when(F.col("b.id1").isNull(), "new in source")
            .when(~F.col("a.name").eqNullSafe(F.col("name1")), "mismatch")
        )
        .withColumn("id", F.coalesce(F.col("id"),F.col("id1")))
        .filter(F.col("comment").isNotNull())
        .select("id","comment")
    )

solve(source_df,target_df).display()
```
