# 36. Distinct daily product sales

**Medium**

Return sell_date, products (array of distinct non-null product names), and null_sell (original output spelling, meaning the count of distinct products). Product array order does not matter.


```python
%run ./_setup $exercise_id="36"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.

    return (
        input_df
        .filter(F.col("product").isNotNull())
        .groupBy("sell_date")
        .agg(
            F.collect_set("product").alias("products"),
            F.count_distinct("product").alias("null_sell")
            )
    )

solve(input_df).display()
```
