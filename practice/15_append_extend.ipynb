# 15. Python append versus extend

**Easy**

This original scenario is a Python-list task, so its inputs and expected output are native Python objects. Using separate copies of l1, return a dict with appended (l2 added as one element) and extended (l2 elements added individually). The two operations must start from the same original l1, not run sequentially on the same list.


```python
%run ./_setup $exercise_id="15"
```

**Your solution:** add imports and return your result from `solve`.


```python
def solve(l1, l2):
    appended = l1.copy()
    extended = l1.copy()

    appended.append(l2)
    extended.extend(l2)

    return {
        "appended":appended,
        "extended":extended,
    }

print(solve(l1,l2))
```
