# 39. Running transaction balance

**Medium**

Keep input columns and add running_total: row-by-row cumulative amount per customer_id ordered by transaction_date, then transaction_id ascending. transaction_id is unique. Include refunds (negative amounts). The upstream heading was copied from the previous scenario; this task is a running balance.


```python
%run ./_setup $exercise_id="39"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window 

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .withColumn(
            "running_total",
            F.sum("amount")
            .over(
                Window.partitionBy("customer_id")
                .rowsBetween(Window.unboundedPreceding,Window.currentRow)
                .orderBy(F.col("transaction_date"),F.col("transaction_id"))
            )
        )
    )
solve(input_df).display()
```
