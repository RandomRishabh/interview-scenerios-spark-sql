# 38. Second distinct salary per department

**Medium**

Return department, emp_name, salary for every employee at the second-highest distinct salary within their department. Include ties. Omit departments with only one salary level. This states the intended salary-ranking rule explicitly.


```python
%run ./_setup $exercise_id="38"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .withColumn("rank", F.dense_rank().over(Window.partitionBy("department").orderBy(F.col("salary").desc())))
        .filter(F.col("rank")==2)
        .drop("emp_id","rank")
    )
solve(input_df).display()
```
