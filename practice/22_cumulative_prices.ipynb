# 22. Cumulative purchase totals

**Medium**

Within each pid, calculate a row-by-row cumulative sum of price ordered by date chronologically. Dates use d-MMM and all belong to 2024; dates are unique per pid. Keep input columns and add new_price. This explicitly uses calendar order rather than the reference ordering by price.


```python
%run ./_setup $exercise_id="22"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

w=(
    Window.partitionBy("pid")
    .orderBy("date_")
    .rowsBetween(Window.unboundedPreceding,Window.currentRow)
)

def solve(input_df):
    # Write your solution here.
    result_df=(
        input_df
        .withColumn("date_", F.to_date(F.concat(F.col("date"),F.lit("-2024")),"d-MMM-yyyy"))
        .withColumn("new_price", F.sum("price").over(w))
        .drop("date_")
    )
    return result_df

solve(input_df).display()
```
