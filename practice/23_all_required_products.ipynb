# 23. Customers buying every required product

**Hard**

Return each customer_id who purchased every distinct product_key in products. Ignore repeated purchases and purchases outside the required catalog. If products is empty, every distinct customer in purchases qualifies. Use the data, never hard-code customer IDs.


```python
%run ./_setup $exercise_id="23"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(purchases, products):
    # Write your solution here.
    result_df=(
        purchases
        .groupBy("customer_id")
        .agg((F.collect_set("product_key")).alias("products_bought"))
        .crossJoin(
            products
            .groupBy()
            .agg((F.collect_set("product_key")).alias("total_products"))
            )
        .filter(
            F.size(F.array_except("total_products","products_bought"))==0
        )
        .select("customer_id")
    )
    return result_df

solve(purchases,products).display()
```
