# 02. Dispatch records for ordered customers

**Easy**

Return every row with status exactly dispatched whose orderid has an Ordered row anywhere in this input. Match statuses case-sensitively. This follows the original sample: consecutive dispatched rows are all returned; this is not a first-transition-only task.


```python
%run ./_setup $exercise_id="2"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.
    return (
        input_df.alias("a")
        .join(input_df.alias("b"),"orderid","left")
        .filter((F.col("a.status")=="dispatched")
                &
                (F.col("b.status")=="Ordered")
                )
        .select("a.orderid","a.statusdate","a.status")
    )
solve(input_df).display()
```
