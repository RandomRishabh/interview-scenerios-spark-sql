**Easy**

Add total as the sum of telugu, english, maths, science and social. All marks are non-null. Keep every input column.

```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.

    return (
        input_df
        .withColumn("total",F.col("telugu") + F.col("english") + F.col("maths") + F.col("science") + F.col("social"))
    )

solve(input_df).display()
```
