# 37. Latest order per customer

**Medium**

Return exactly one latest order per customer_id by order_date (ISO yyyy-MM-dd strings). If two orders share the latest date, choose the higher order_id. order_id is unique and dates are non-null. Keep order_id, customer_id, order_date, amount. The tie rule makes the original exercise deterministic.


```python
%run ./_setup $exercise_id="37"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.

    return (
        input_df
        .withColumn("order_date1",F.to_date("order_date","yyyy-MM-dd"))
        .withColumn("rank", F.row_number().over(Window.partitionBy("customer_id").orderBy(F.col("order_date1").desc(),F.col("order_id").desc())))
        .filter(F.col("rank")==1)
        .drop("rank","order_date1")
    )

solve(input_df).display()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.

    return (
        input_df
        .groupBy("customer_id")
        .agg(
            F.max(
                F.struct("order_date","order_id","amount")
            ).alias("latest")
        )
        .select("customer_id","latest.*")
    )

solve(input_df).display()
```
