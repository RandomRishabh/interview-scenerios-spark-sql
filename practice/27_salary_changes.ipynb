# 27. Salary change from previous observation

**Medium**

Within empid, order by year and add incresalary (original spelling): current salary minus previous observed salary, or 0 for the first observation. Years are unique per employee; gaps are allowed and decreases must stay negative.


```python
%run ./_setup $exercise_id="27"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .withColumn("incresalary", F.col("salary") - F.lag("salary").over(Window.partitionBy("empid").orderBy("year"))).fillna({"incresalary":0})
    )
solve(input_df).display()
```
