# 17. Merge complementary employee datasets

**Medium**

Combine the employee information in df1 and df2. Records describe the same employee when emp_id, name and age all match. Input keys are unique and non-null. Return emp_id, name, age, state, country, address, preserving unmatched rows with nulls.


```python
%run ./_setup $exercise_id="17"
```

**Your solution:** add imports and return your result from `solve`.


```python
def solve(df1, df2):
    # Write your solution here.
    result_df= (
        df1.alias("a")
        .join(
            df2.alias("b"),
            on=["emp_id","name","age"],
            how="outer"
        )
    )

    return result_df

solve(df1,df2).display()
```
