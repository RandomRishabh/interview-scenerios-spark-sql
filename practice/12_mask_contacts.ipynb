# 12. Mask contact details

**Medium**

Mask each non-null email as its first character + exactly ten asterisks + the substring starting at zero-based index 8. Mask mobile as its first two characters + exactly five asterisks + its last three characters. Preserve nulls. These positional rules intentionally match the original exercise; they are not a production masking policy.

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    
    result_df= (
        input_df
        .withColumn("email_name",F.split_part("email",F.lit("@"),F.lit(1)))
        .withColumn("email_domain",F.split_part("email",F.lit("@"),F.lit(2)))
        .withColumn(
            "email_name_mask", 
            F.concat(
                (F.substring("email_name",1,1))
                ,F.lit("**********")
                ,F.col("email_name").substr(F.lit(9),F.length("email_name"))
                )
            )
        .withColumn("email",F.concat(F.col("email_name_mask"),F.lit("@"),F.col("email_domain")))
        .withColumn(
            "mobile",
            F.concat(
                (F.substring("mobile",1,2))
                ,F.lit("*****")
                ,(F.substring("mobile",-3,3))
            )
        )
        .select("email","mobile")
    )

    return result_df

solve(input_df).display()
```
