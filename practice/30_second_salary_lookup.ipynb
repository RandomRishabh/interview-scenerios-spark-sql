# 30. Second distinct salary with department lookup

**Hard**

employees.dept_id identifies the department listed as departments.dept_id1. Return every employee with the second-highest distinct salary in their department, including salary ties. Departments with fewer than two salary levels yield no rows. Return emp_id, name, dept_name, salary.


```python
%run ./_setup $exercise_id="30"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(employees, departments):
    # Write your solution here.

    return (
        employees
        .withColumn("rank", F.dense_rank().over(Window.partitionBy("dept_id").orderBy(F.col("salary").desc())))
        .filter(F.col("rank")==2)
        .join(departments,employees.dept_id==departments.dept_id1,"inner")
        .select("emp_id","name","dept_name","salary")
    )

solve(employees,departments).display()
```
