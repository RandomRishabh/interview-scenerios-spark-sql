# 25. Parse and quarantine malformed CSV records

**Hard**

raw_lines contains CSV records, including a header row that can occur anywhere. Parse standard CSV quoting, skip the header, and return emp_no, emp_name, dep as strings. Reject records with a field count other than three. Preserve quoted commas. This portable fixture checks output correctness; discuss ingestion-time rejection separately rather than writing to DBFS.

```python
import csv
from pyspark.sql import functions as F

def solve(raw_lines):

    def parse_line(line):
        if line is None:
            return None
        
        try:
            fields = next(csv.reader([line], strict=True))
        except (csv.Error, StopIteration):
            return None
        
        if len(fields) !=3:
            return None
        
        if fields == ["emp_no","emp_name","dep"]:
            return None
        
        return fields
    
    parse_csv= F.udf(parse_line, "array<string>")

    return (
        raw_lines
        .withColumn("fields", parse_csv("raw_line"))
        .filter(F.col("fields").isNotNull())
        .select(
            F.col("fields")[0].alias("emp_no"),
            F.col("fields")[1].alias("emp_name"),
            F.col("fields")[2].alias("dep")
        )
    )

solve(raw_lines).display()
```
