# 42. Duplicate transaction counts

**Easy**

Return transaction_id and occurrence_count for IDs appearing more than once. IDs are non-null. Count input rows, including exact repeats. Use the correctly spelled occurrence_count.


```python
%run ./_setup $exercise_id="42"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .groupBy("transaction_id")
        .agg(F.count("transaction_id").alias("occurrence_count"))
        .filter(F.col("occurrence_count")>1)
    )

solve(input_df).display()
```
