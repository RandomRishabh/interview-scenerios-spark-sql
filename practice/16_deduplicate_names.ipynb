# 16. Deterministic deduplication by name

**Deduplication and windows · Medium**

Keep exactly one row per name: the row with the smallest id. IDs are unique. Keep all input columns. This makes the original dropDuplicates example deterministic.


```python
%run ./_setup $exercise_id="16"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .withColumn("rank",F.row_number().over(Window.partitionBy("name").orderBy(F.col("id"))))
        .filter(F.col("rank")==1)
        .drop("rank")
    )

solve(input_df).display()
```
