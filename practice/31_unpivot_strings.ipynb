# 31. Convert string columns into rows

**Medium**

Turn col1, col2, col3 and col4 into separate rows in one column named col. Keep duplicates. These inputs contain non-null strings. Commas inside strings remain unchanged; do not add an extra blank row (the upstream printed table had a stray blank).


```python
%run ./_setup $exercise_id="31"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .select(
            F.explode(F.array("col1","col2","col3","col4"))
        ).alias("col")
    )

solve(input_df).display()
```
