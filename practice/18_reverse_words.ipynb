# 18. Reverse each word

**Easy**

Reverse characters inside each space-separated word while retaining word order and spaces. Return one column named reverse word. Preserve null and empty strings.


```python
%run ./_setup $exercise_id="18"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F


def solve(input_df):
    # Write your solution here.
    result_df = input_df.withColumn(
        "reverse word",
        F.array_join(
            F.transform(
                F.split("word", " "), lambda word: F.reverse(word)
                )
            , " "
        ),
    ).drop("word")
    return result_df


solve(input_df).display()
```
