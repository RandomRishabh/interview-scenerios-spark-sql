# 07. Highest quantity sales per year

**Windows and ties · Medium**

Return all sales with the maximum quantity within their year. Include ties and keep the input columns.


```python
%run ./_setup $exercise_id="7"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):

    result_df=(
        input_df
        .withColumn(
            "max_quantity",
            F.max("quantity").over(Window.partitionBy("year"))
        )
        .filter(F.col("quantity")==F.col("max_quantity"))
        .orderBy("year")
        .drop("max_quantity")

    )
    
    return result_df

solve(input_df).show()
```
