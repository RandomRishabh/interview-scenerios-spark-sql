# 08. Unique team matchups

**Medium**

Return one matches string per unordered pair of distinct teams, formatted smaller-name Vs larger-name using lexical order. Input team names are unique.


```python
%run ./_setup $exercise_id="8"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F

def solve(input_df):
    # Write your solution here.
    return (
        input_df.alias("a")
        .crossJoin(input_df.alias("b"))
        .filter(
            (F.col("a.teams")<F.col("b.teams"))
            )
        .withColumn("matches",F.concat(F.col("a.teams"),F.lit(" Vs "),F.col("b.teams")))
        .select("matches")
    )

solve(input_df).show()
```
