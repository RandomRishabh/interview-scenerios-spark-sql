# 34. Customer age distribution

**Easy**

Group ages into 19-35 inclusive, 36-50 inclusive, 51+ (age >=51), and Other (under 19 or null). Return age_group, count. Corrected boundary: 51 belongs to 51+.


```python
%run ./_setup $exercise_id="34"
```

**Your solution:** add imports and return your result from `solve`.


```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def solve(input_df):
    # Write your solution here.
    return (
        input_df
        .withColumn(
            "age_group", 
            F.when((F.col("age")<= 35) & (F.col("age")>=19), "19-35")
            .when((F.col("age")<= 50) & (F.col("age")>=36), "36-50")
            .when((F.col("age")>=51) , "51+")
            .otherwise("Other")
            )
        # .filter(F.col("age_group")!="other")
        .groupBy("age_group")
        .agg(F.count("customer_id").alias("count"))
    )

solve(input_df).display()
```
